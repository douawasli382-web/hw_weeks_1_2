### Exercice 2 – Lien entre partie réelle et imaginaire
Soit $z = e^{j\theta}$. Montrer que
$$ \cos\theta = \frac{z+z^{-1}}{2} \quad\text{et}\quad \sin\theta = \frac{z-z^{-1}}{2j}. $$
En déduire une expression de $\cos^3\theta$ en fonction de $\cos(3\theta)$ et $\cos\theta$ (formule de triple angle). Utiliser cette relation pour exprimer $\cos^3\theta$ comme combinaison linéaire de cosinus.

**Démonstration** (à compléter) :
**Démonstration :**

**1. Formules d'Euler :**
Soit $z = e^{i\theta}$. On a $z^{-1} = e^{-i\theta}$.
D'après la formule d'Euler :
$$z + z^{-1} = e^{i\theta} + e^{-i\theta} = (\cos\theta + i\sin\theta) + (\cos\theta - i\sin\theta) = 2\cos\theta \implies \cos\theta = \frac{z + z^{-1}}{2}$$

$$z - z^{-1} = e^{i\theta} - e^{-i\theta} = (\cos\theta + i\sin\theta) - (\cos\theta - i\sin\theta) = 2i\sin\theta \implies \sin\theta = \frac{z - z^{-1}}{2j}$$

**2. Linéarisation de $\cos^3\theta$ :**
En utilisant la formule d'Euler :
$$\cos\theta = \frac{e^{i\theta} + e^{-i\theta}}{2}$$

En élevant au cube :
$$\cos^3\theta = \left(\frac{e^{i\theta} + e^{-i\theta}}{2}\right)^3 = \frac{1}{8} \left( e^{i3\theta} + 3 e^{i2\theta} e^{-i\theta} + 3 e^{i\theta} e^{-i2\theta} + e^{-i3\theta} \right)$$

$$\cos^3\theta = \frac{1}{8} \left( (e^{i3\theta} + e^{-i3\theta}) + 3(e^{i\theta} + e^{-i\theta}) \right)$$

Comme $e^{i3\theta} + e^{-i3\theta} = 2\cos(3\theta)$ et $e^{i\theta} + e^{-i\theta} = 2\cos\theta$ :
$$\cos^3\theta = \frac{1}{8} \left( 2\cos(3\theta) + 6\cos\theta \right)$$

On obtient la combinaison linéaire :
$$\cos^3\theta = \frac{3}{4}\cos\theta + \frac{1}{4}\cos(3\theta)$$

```python
# Vérification numérique de la formule
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, 2*np.pi, 100)
cos3 = np.cos(theta)**3

# Utiliser la formule dérivée pour vérifier
cos3_formule = (3/4) * np.cos(theta) + (1/4) * np.cos(3*theta)

# Tracer les deux pour vérifier
plt.plot(theta, cos3, label='cos^3')
plt.plot(theta, cos3_formule, '--', label='formule')
plt.legend()
plt.grid()
plt.show()
```

### Exercice 3 – Équation avec exponentielle
Trouver tous les réels $\theta$ tels que $e^{j\theta} + e^{-j\theta} = \sqrt{3}$.

Que valent $\cos\theta$ et $\sin\theta$ ?

**Solution** (à compléter) :
**1. Calcul de $\cos\theta$ et $\sin\theta$ :**

D'après la formule d'Euler :
$$e^{j\theta} + e^{-j\theta} = 2\cos\theta$$

L'équation donnée est :
$$e^{j\theta} + e^{-j\theta} = \sqrt{3}$$

Ce qui donne :
$$2\cos\theta = \sqrt{3} \implies \cos\theta = \frac{\sqrt{3}}{2}$$

Pour $\sin\theta$, on utilise l'identité $\cos^2\theta + \sin^2\theta = 1$ :
$$\sin^2\theta = 1 - \cos^2\theta = 1 - \left(\frac{\sqrt{3}}{2}\right)^2 = 1 - \frac{3}{4} = \frac{1}{4}$$

Donc :
$$\sin\theta = \pm \frac{1}{2}$$

---

**2. Recherche des réels $\theta$ :**

* **Cas 1 :** $\cos\theta = \frac{\sqrt{3}}{2}$ et $\sin\theta = \frac{1}{2}$
  $$\theta = \frac{\pi}{6} + 2k\pi \quad (k \in \mathbb{Z})$$

* **Cas 2 :** $\cos\theta = \frac{\sqrt{3}}{2}$ et $\sin\theta = -\frac{1}{2}$
  $$\theta = -\frac{\pi}{6} + 2k\pi \quad (k \in \mathbb{Z})$$

**Conclusion :**
Les solutions sont :
$$\theta = \pm \frac{\pi}{6} + 2k\pi \quad (k \in \mathbb{Z})$$

### Exercice 6 – Battements
Deux signaux de fréquences proches :  
$
x_1(t) = \cos(2\pi f_1 t)
$ et
$
x_2(t) = \cos(2\pi f_2 t)
$.  
Montrer que leur somme peut s’écrire  
$$
2\cos(2\pi f_m t)\cos(2\pi f_d t)
$$  
avec $f_m = \frac{f_1+f_2}{2}$ et $f_d = \frac{f_1-f_2}{2}$.  
Pour $f_1 = 10$ Hz et $f_2 = 12$ Hz, tracer la somme et observer l’enveloppe. Expliquer le phénomène de battement.

**Démonstration** (à compléter) :
**1. Démonstration de la formule des battements :**
Soit la somme des deux signaux :
$$x(t) = x_1(t) + x_2(t) = \cos(2\pi f_1 t) + \cos(2\pi f_2 t)$$

En utilisant la formule trigonométrique de transformation somme-produit :
$$\cos(a) + \cos(b) = 2 \cos\left(\frac{a+b}{2}\right) \cos\left(\frac{a-b}{2}\right)$$

En posant $a = 2\pi f_1 t$ et $b = 2\pi f_2 t$ :
$$x(t) = 2 \cos\left(\frac{2\pi f_1 t + 2\pi f_2 t}{2}\right) \cos\left(\frac{2\pi f_1 t - 2\pi f_2 t}{2}\right)$$
$$x(t) = 2 \cos\left(2\pi \frac{f_1 + f_2}{2} t\right) \cos\left(2\pi \frac{f_1 - f_2}{2} t\right)$$

En introduisant les fréquences $f_m = \frac{f_1 + f_2}{2}$ (fréquence moyenne) et $f_d = \frac{f_1 - f_2}{2}$ (fréquence de différence/battement) :
$$x(t) = 2 \cos(2\pi f_m t) \cos(2\pi f_d t)$$

**2. Explication du phénomène de battement :**
Le signal représente une oscillation rapide à la fréquence moyenne $f_m$ dont l'amplitude est modulée lentement par le terme $\cos(2\pi f_d t)$. On observe des variations périodiques de l'amplitude (des gonflements ou "battements") à la fréquence $f_d$.

```python
f1 = 10
f2 = 12
t = np.linspace(0, 1, 2000)

x1 = np.cos(2*np.pi*f1*t)
x2 = np.cos(2*np.pi*f2*t)
x_sum = x1 + x2

fm = (f1+f2)/2
fd = (f1-f2)/2
x_env = 2*np.cos(2*np.pi*fd*t)  # enveloppe
x_product = x_env * np.cos(2*np.pi*fm*t)

plt.plot(t, x_sum, label='Somme')
plt.plot(t, x_env, '--', label='Enveloppe')
plt.plot(t, -x_env, '--', label='-Enveloppe')
plt.xlabel('Temps (s)')
plt.legend()
plt.grid()
plt.show()

# Explication : la somme de deux sinusoïdes de fréquences proches donne une oscillation rapide (fm) modulée par une enveloppe lente (fd).
```

Générer ce signal pour $f_c = 100$ Hz, $f_m = 5$ Hz, $m = 0.5$, $A_c = 1$.  
Tracer le signal sur 0,2 s, puis calculer sa transformée de Fourier (FFT) pour visualiser les raies spectrales (porteuse et bandes latérales).  
Expliquer pourquoi il y a trois composantes fréquentielles.

**Code** :
**Explication des 3 composantes fréquentielles :**

En développant l'expression du signal modulé $s(t)$ :
$$s(t) = A_c [1 + m \cos(2\pi f_m t)] \cos(2\pi f_c t)$$

En utilisant la trigonométrie, on trouve trois composantes fréquentielles dans le spectre :
1. **La porteuse :** à la fréquence $f_c = 100\text{ Hz}$.
2. **La bande latérale supérieure (USB) :** à la fréquence $f_c + f_m = 105\text{ Hz}$.

```python
fs = 1000  # Hz
t = np.linspace(0, 0.2, int(0.2*fs))
fc = 100
fm = 5
m = 0.5
Ac = 1

s = Ac * (1 + m*np.cos(2*np.pi*fm*t)) * np.cos(2*np.pi*fc*t)

plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
plt.plot(t, s)
plt.xlabel('Temps (s)')
plt.title('Signal modulé en amplitude')
plt.grid()

# FFT
N = len(s)
freqs = np.fft.fftfreq(N, 1/fs)
spectre = np.fft.fft(s)

plt.subplot(1, 2, 2)
plt.stem(freqs[:N//2], np.abs(spectre[:N//2]) / N)
plt.xlim(0, 150)
plt.xlabel('Fréquence (Hz)')
plt.ylabel('Amplitude')
plt.title('Spectre (3 raies)')
plt.grid()
plt.show()

# Explication : le produit de cosinus donne des composantes à fc, fc+fm, fc-fm.
```

### Exercice 9 – Repliement de spectre (aliasing) – approche numérique
Échantillonner le signal $x(t) = \cos(2\pi 30 t)$ avec une fréquence d’échantillonnage $f_e = 20$ Hz (donc sous-échantillonnage).  
Générer les échantillons pour $t = 0, 0.05, 0.10, \dots$ et les tracer.  
Reconstruire le signal à partir des échantillons en traçant la droite reliant les points (ou en utilisant une interpolation).  
Quelle fréquence apparente observez-vous ? Justifier à l’aide de la formule d’aliasing.

**Code** :
**Justification théorique (Formule d'aliasing) :**
* Le signal d'origine a une fréquence $f = 30\text{ Hz}$.
* La fréquence d'échantillonnage est $f_e = 20\text{ Hz}$.
* La fréquence de Nyquist est $\frac{f_e}{2} = 10\text{ Hz}$. Puisque $f > \frac{f_e}{2}$, on est en situation de **sous-échantillonnage** (repliement de spectre / aliasing).

La fréquence apparente (ou repliée) $f_{app}$ se calcule par la formule :
$$f_{app} = |f - k \cdot f_e|$$
En choisissant $k = 1$ (pour ramener la fréquence dans l'intervalle $[-\frac{f_e}{2}, \frac{f_e}{2}]$ ou $[0, f_e]$) :
$$f_{app} = |30 - 1 \times 20| = 10\text{ Hz}$$
*(Ou bien par la formule $|30 - 20| = 10\text{ Hz}$)*.
On observe donc une fréquence apparente de **10 Hz**.

```python
fe = 20
f = 30
t_cont = np.linspace(0, 0.5, 1000)
x_cont = np.cos(2*np.pi*f*t_cont)

# Échantillonnage
te = 1/fe
t_ech = np.arange(0, 0.5, te)
x_ech = np.cos(2*np.pi*f*t_ech)

plt.plot(t_cont, x_cont, label='Signal original (30 Hz)')
plt.stem(t_ech, x_ech, linefmt='r-', markerfmt='ro', basefmt=' ', label='Échantillons (20 Hz)')

# Reconstruction linéaire (simple interpolation)
t_recon = np.linspace(0, 0.5, 1000)
x_recon = np.interp(t_recon, t_ech, x_ech)
plt.plot(t_recon, x_recon, '--', label='Reconstruction')

plt.legend()
plt.grid()
plt.show()

# La fréquence apparente est |fe - f| = 10 Hz. En effet, l'échantillonnage à 20 Hz donne un alias à 10 Hz.
```

### Exercice 10 – Synthèse d’un signal carré par séries de Fourier
Un signal carré de période $T$ et d’amplitude 1 peut être décomposé en série de Fourier :  
$$
x(t) = \frac{4}{\pi} \sum_{k=0}^{N} \frac{1}{2k+1} \sin\bigl(2\pi (2k+1) f_0 t\bigr).
$$  
Écrire une fonction Python qui génère ce signal pour un nombre d’harmoniques $N$ donné.  
Tester pour $N=1, 5, 20, 50$ et tracer les signaux correspondants.  
Observer le phénomène de Gibbs (oscillations près des discontinuités).  
Comparer avec la fonction `square` de `scipy.signal`.

**Code** :
**Observations et Réponses :**
* **Effet de N :** Plus le nombre d'harmoniques $N$ augmente (de 1 à 50), plus la somme de Fourier s'approche de la forme d'un signal carré idéal.
* **Phénomène de Gibbs :** On observe des dépassements (oscillations) au niveau des fronts de montée et de descente du signal. Ces oscillations persistent même pour de grandes valeurs de $N$.
* **Comparaison avec SciPy :** La fonction `scipy.signal.square` génère un signal carré parfait avec des transitions instantanées, contrairement à la série de Fourier qui reste une approximation polynomiale trigonométrique.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

def carre_fourier(t, f0, N):
    """Génère un signal carré par synthèse de Fourier avec N harmoniques impaires."""
    x = np.zeros_like(t)
    for k in range(N):
        n = 2*k + 1
        x += (1/n) * np.sin(2*np.pi*n*f0*t)
    return (4/np.pi) * x

f0 = 1  # Hz
t = np.linspace(0, 2, 1000)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
Ns = [1, 5, 20, 50]

for ax, N in zip(axes.flat, Ns):
    x = carre_fourier(t, f0, N)
    ax.plot(t, x, label=f'N={N}')
    ax.grid()
    ax.legend()

plt.suptitle('Synthèse d\'un signal carré par séries de Fourier')
plt.tight_layout()
plt.show()

# Comparaison avec square
x_square = signal.square(2*np.pi*f0*t)
plt.plot(t, x_square, label='square (scipy)')
plt.plot(t, carre_fourier(t, f0, 50), '--', label='N=50')
plt.legend()
plt.grid()
plt.show()
```

### Exercice 11 – Phase et décalage temporel
Montrer que $A\cos(2\pi f t - \phi)$ est équivalent à un décalage temporel. Quel est le décalage en secondes ?  
Pour $A=1, f=2$ Hz, $\phi = \pi/2$, calculer le décalage. Tracer le signal non décalé et le signal décalé pour vérifier.

**Démonstration** : $\cos(2\pi f (t - \Delta t)) = \cos(2\pi f t - 2\pi f \Delta t)$ donc $\phi = 2\pi f \Delta t$ et $\Delta t = \phi/(2\pi f)$.

**Code** :
**Application numérique :**
* Fréquence $f = 2\text{ Hz}$
* Phase $\phi = \frac{\pi}{2}\text{ rad}$

Le décalage temporel $\Delta t$ se calcule par la formule :
$$\Delta t = \frac{\phi}{2\pi f} = \frac{\pi / 2}{2\pi \times 2} = \frac{\pi / 2}{4\pi} = \frac{1}{8}\text{ s} = 0.125\text{ s}$$

Le signal décalé est donc en avance (ou en retard selon le signe) de $0.125\text{ secondes}$ par rapport au signal de référence $A\cos(2\pi f t)$.

```python
f = 2
phi = np.pi/2
A = 1
delta_t = phi / (2*np.pi*f)
print(f"Décalage temporel : {delta_t:.4f} s")

t = np.linspace(0, 1, 1000)
x_original = A * np.cos(2*np.pi*f*t)
x_decale = A * np.cos(2*np.pi*f*(t - delta_t))
x_phase = A * np.cos(2*np.pi*f*t - phi)

plt.plot(t, x_original, label='Original (phi=0)')
plt.plot(t, x_decale, '--', label='Décalé temporel')
plt.plot(t, x_phase, ':', label='Phase -π/2')
plt.legend()
plt.grid()
plt.show()
```

### Exercice 12 – Échantillonnage d’une exponentielle complexe
Soit $z[n] = e^{j\omega_0 n}$. Montrer que si $\omega_0$ est un multiple rationnel de $2\pi$, le signal est périodique.  
Pour $\omega_0 = 2\pi/3$, déterminer la période fondamentale.  
Pour $\omega_0 = 1$ rad/échantillon, le signal est-il périodique ? Justifier.  
Générer les deux signaux sur 50 échantillons et observer leur comportement.

**Démonstration** : La périodicité exige $\omega_0 N = 2\pi k$ avec $k$ entier, donc $\omega_0/(2\pi) = k/N$ est rationnel.

**Code** :
**Applications numériques :**
1. **Pour $\omega_0 = \frac{2\pi}{3}$ :**
   $$\frac{\omega_0}{2\pi} = \frac{2\pi / 3}{2\pi} = \frac{1}{3}$$
   C'est un nombre rationnel ($k=1, N=3$). Le signal est donc **périodique** de période fondamentale $N = 3$ échantillons.

2. **Pour $\omega_0 = 1\text{ rad/échantillon}$ :**
   $$\frac{\omega_0}{2\pi} = \frac{1}{2\pi}$$
   Ce n'est pas un nombre rationnel (car $\pi$ est irrationnel). Le signal n'est **pas périodique**.

```python
n = np.arange(0, 50)

omega1 = 2*np.pi/3
z1 = np.exp(1j * omega1 * n)

omega2 = 1
z2 = np.exp(1j * omega2 * n)

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.stem(n, np.real(z1), basefmt=' ')
plt.title(f'$\\omega_0 = 2\pi/3$ (périodique, période 3)')

plt.subplot(1, 2, 2)
plt.stem(n, np.real(z2), basefmt=' ')
plt.title('$\\omega_0 = 1$ (non périodique)')

plt.tight_layout()
plt.show()

# Vérification : pour omega1, la période est 3 car 2pi/3 * 3 = 2pi.
```
